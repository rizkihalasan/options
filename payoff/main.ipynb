## Strategy:
### B. Spread

#### B.1. Bull spreads

```python
options1 = Option('call', 'long', 100, 2.7, 'Mar 26', 1, None, quantity=1)
options2 = Option('call', 'short', 105, 1, 'Mar 26', 1, None, quantity=1)

options = [options1, options2]
s = Strategy(options, None)
s.calculate_profit()
```

#### B.2. Bear spreads

```python
options1 = Option('put', 'long', 100, 2.7, 'Mar 26', 1, None, quantity=1)
options2 = Option('put', 'short', 95, 1, 'Mar 26', 1, None, quantity=1)

options = [options1, options2]
s = Strategy(options, None)
s.calculate_profit()
```

#### B.3. Box spreads

```python
options1 = Option('call', 'long', 100, 2.7, 'Mar 26', 1, None, quantity=1)
options2 = Option('call', 'short', 105, 1, 'Mar 26', 1, None, quantity=1)
options3 = Option('put', 'long', 100, 2.7, 'Mar 26', 1, None, quantity=1)
options4 = Option('put', 'short', 105, 6, 'Mar 26', 1, None, quantity=1)

options = [options1, options2, options3, options4]
s = Strategy(options, None)
s.calculate_profit()
```

#### B.4. Butterfly Spreads

```python
options1 = Option('call', 'long', 80, 25, 'Mar 26', 1, None, quantity=1)
options2 = Option('call', 'long', 100, 2.7, 'Mar 26', 1, None, quantity=1)
options3 = Option('call', 'short', 90, 13.5, 'Mar 26', 1, None, quantity=2)

options = [options1, options2, options3]
s = Strategy(options, None)
s.calculate_profit()
```
